```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
docs = ['go india',
        'india india',
        'hip hip hurray',
        'jeetega bhai jeetega india jeetega',
        'bharat mata ki jai',
        'kohli kohli',
        'sachin sachin',
        'dhoni dhoni',
        'modi ji ki jai',
        'inquilab zindabad']
```

#Tokenizing or integer encoding

```python
from tensorflow.keras.preprocessing.text import Tokenizer
tokenizer=Tokenizer(oov_token='<nothing>')
```

```python
tokenizer.fit_on_texts(docs)
```

#What integer has been assigned to each word

```python
tokenizer.word_index
```

#How many times each word has occurred?

```python
tokenizer.word_counts
```

```python
tokenizer.document_count
```

```python
sequences=tokenizer.texts_to_sequences(docs)
sequences
```

['go india']=[10,2]

#How to make each sequence of same size?

```python
from keras.utils import pad_sequences

sequences=pad_sequences(sequences,padding='post')
sequences
```

#This sequence is what that is put to the RNN as input layer:

```python
from tensorflow import keras
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense

model_d=Sequential()
model_d.add(Embedding(17,output_dim=2,input_length=5))

model_d.add(SimpleRNN(32,return_sequences=False))
model_d.add(Dense(1,'sigmoid'))
```

```python
model_d.compile(
    loss='binary_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)
```

```python
model_d.build(input_shape=(None, 5))
```

```python
model_d.summary()
```

#Let us now do sentiment analysis using RNN

```python
from tensorflow.keras.datasets import imdb
from keras import Sequential
from keras.layers import Dense,SimpleRNN,Embedding,Flatten
```

```python
(x_train,y_train),(x_test,y_test)=imdb.load_data()
```

```python
x_train[0]
```

```python
y_train[0]
```

```python
len(x_train[0])
```

#So it is required to do padding here so that all sequences are of same length

```python
x_train=pad_sequences(x_train,padding='post')
x_test=pad_sequences(x_test,padding='post')
```

```python
len(x_train)
```

```python
max_len_vector = x_train.shape[1]
print(f"The maximum length of a vector in x_train is: {max_len_vector}")
```

```python
print(x_train[14000])
```

```python
model=Sequential()


model.add(Embedding(input_dim=10000,output_dim=2,input_length=2494))
model.add(SimpleRNN(32,return_sequences=False))
model.add(Dense(1,'sigmoid'))

```

```python
model.compile(
    loss='binary_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)
```

```python
model.summary()
```

```python
model.fit(x_train,y_train,epochs=10,validation_data=(x_test,y_test))
```
