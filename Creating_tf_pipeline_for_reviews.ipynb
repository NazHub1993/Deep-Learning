#This is how to create labels

```python
def get_label(file_path):
    import os
    parts=tf.strings.split(file_path,os.path.sep)
    return parts[-2]
```

```python
get_label('/content/drive/MyDrive/Reviews/reviews/negative/neg_1.txt')
```

#Let's divide it into train and test set

```python
rev_cnt=int(len(reviews_ds))
rev_cnt
```

```python
def process_all(file_path):
  label=get_label(file_path)
  content=tf.io.read_file(file_path)
  return content,label
```

```python
reviews_ds_duplicate=reviews_ds
```

```python
reviews_ds=reviews_ds.map(process_all)
```

```python
for content,label in reviews_ds:
  print('Content: ',content.numpy())
  print('Label: ',label.numpy())
```

```python
reviews_ds=reviews_ds.filter(lambda content,label:content!="")
for content,label in reviews_ds:
  print('Content: ',content.numpy())
  print('Label: ',label.numpy())
```

#If you want to peform all the opearations in one line

```python
final_ds=reviews_ds_duplicate.map(process_all).filter(lambda content,label:content!="").shuffle(2)
```

```python
for content,label in final_ds.as_numpy_iterator():
  print('Review: ',content)
  print('Label: ',label)
```
