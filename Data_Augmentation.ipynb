```python
from tensorflow.keras.preprocessing import image
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
img=image.load_img('/content/dog_78.jpg',target_size=(200,200))
img
```

```python
type(img)
```

```python
datagen=ImageDataGenerator(
    rotation_range=30,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    width_shift_range=0.2,
    height_shift_range=0.2,
    fill_mode='nearest'
)
```

#datagen OBJECT is ready!!!

```python
img=image.img_to_array(img)
```

```python
img.shape
```

```python
input=img.reshape(1,200,200,3)
```

```python
i=0

for output in datagen.flow(input,batch_size=1,save_to_dir='augmented'):
  i=i+1

  if i==10:
    break
```

```python
import shutil
shutil.make_archive('aug', 'zip', 'aug')   # 'aug' is your folder name
```
